```python
#Conditional statement 
#if , elif , else
age = int(input("Enter age:"))
if age >= 18:
    print("You can vote")
else: 
     print("You cant vot")
    
```

```python
#traffic light 

color = input("Enter color:")

if  color == "red":
    print("Stop")
elif color == "green":
        print("go")
elif color == "yellow":
        print("look")
else :
    print("Wrong color  for traffic light")
```

```python
#teenage or  adult
age = int(input("Enter your age :"))

if (age < 13):
          print("Child")
elif(age >= 13 and age < 18):
    print("Teenage")
else:
      print("adult")
    
```

```python
#login page  
username = input("Enter your username :")
password = input("Enter your password:")

if(username ==  "admin"  and  password =="pass"):
    print("Login successful!")

elif (username != "admin"):
    print("Wrong username")

else:
    print("Wrong password")
```

```python
#if n is mutiple of 5 or not 

number = int(input("Enter a no :"))

if(number%5 == 0):
    print("The  no is multiple by 5")

else:
    print("Not multiple of 5")
    
```

```python
#print if any number is odd or Even 

n = int(input("Enter a no :"))

if(n%2==0):
    print("EVEN")
else:
    print("ODD")
```
