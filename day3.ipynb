```python
#how can we count a no of character

word = "artificial inteligence"

#count the no of i's 

count = 0

for ch in word:
    if(ch == 'i'):
        count+=1
print("count of i is :", count)

        
```

```python
#examples
#print vowel count  of given string

a = str(input("Enter a string:"))

count = 0

for ch in a:
    if(ch == 'i' or  ch == 'e' or  ch == 'a' or  ch == 'o' or ch == 'u'):
        count += 1

print("print the count of the vowel:", count)
     
```
