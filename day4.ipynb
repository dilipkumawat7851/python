```python
#example
def sum(a,b):#parameter and a and b there are a placeholder 
    s = a + b
    return s

#function call 
ans = sum(3,4)
print(ans)
```

```python
# take 3 input and give the abg of the three

#calc average 

def calc_avg(a,b,c):
    sum = a + b + c
    return sum/3

print(calc_avg(1,2,4))
```

```python
#if there was a value is default 
def cal(a,b=1):
    sum = a + b
    return sum

print(cal(1))
```
