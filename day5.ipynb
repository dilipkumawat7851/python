````python
#write a function to print factorial of 'n'
#n!= 1*2*3*4........*n

# n = int(input("Enter number:"))

def cal_factorial(n):
    fact = 1
    for i in range(1,n+1):
        fact *= i
        
    return fact

n = int(input("Enter number:"))
print(cal_factorial(n))

````

````python
# Q8. Let's create a Simple Calculator that performs arithmetic operations. Create
# a function
# calculator(a, b, operation) that performs addition, subtraction,
# multiplication, or division based on the operat ion
# parameter.
# [ operation parameter can have value


# ```python

def calculator(a, b, operation):


    if operation == "add":
        return a + b

    elif operation == "sub":
        return a - b

    elif operation == "multi":
        return a * b

    elif operation == "divide":
        return a / b

    else:
        return "Invalid operation"


a = float(input("Enter a no: "))
b = float(input("Enter second no: "))

operation = input("Enter operation (add, sub, multi, divide): ")

print(calculator(a, b, operation))
````

````python
# Q9. Write a function that returns True if n is a prime number and
# False
# otherwise, using a loop.

def is_prime(n):
    if n < 2:
        return False

    if n == 2:
        return True

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


n = int(input("Enter a number: "))
print(is_prime(n))
````
