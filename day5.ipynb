```python
#write a function to print factorial of 'n'
#n!= 1*2*3*4........*n

# n = int(input("Enter number:"))

def cal_factorial(n):
    fact = 1
    for i in range(1,n+1):
        fact *= i
        
    return fact

n = int(input("Enter number:"))
print(cal_factorial(n))

```
